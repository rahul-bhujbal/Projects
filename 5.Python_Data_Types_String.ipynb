```python
Strings are sequance of character

In Python specifically, Strings are sequance of unicode character

Creating String
Accessing String
Adding character in string
Editing String
Deleting String
Operation on String
String Functions
```

```python
# Create String

s = 'hello'
print(s)
s = "hello"
print(s)
# Multiline string
s = '''hello'''
print(s)
s = """ hello"""
print(s)
s = str('hello')

print(s)
```

```python
# why single and double quato in python

s = 'Its most imp'
print(s)

s = "It's most imp"
print(s)

s = """Python is very intrested language,
and it's also so easy"""

print(s)

s = '''Python is very intrested language,
and its also so easy'''

print(s)
```

```python
# Accessing String

s = 'hello world'

# Indexing

print(s[0])

print(s[6])
print(s[5])

# negative indexing
print(s[-1])
```

```python
# slicing
# Positive Indexing
s = 'hello world'
print(s[0:5])
print(s[2:3])
print(s[2:])
print(s[:3])
print(s[:])
print(s[0:6:2])
print(s[0:10:3])
```

```python
# Nigetaive Indexing


s = 'hello world'
print(s[6:0:-1])

# Reverse String
print(s[::-1])

print(s[-5:])
```

```python
# Editing & deleting string

s = 'hello world'

print(s[0])
# s[0]='H'
# Python strings are immutable

s = 'hello world'

del s

```

```python
# Operations on string
# Arithmetaic Operation
print('Pune' + 'Mumbai')
print('Pune' +' ' + 'Mumbai')
#print('Pune' - 'Mumbai')
print('Pune'*5)
print('*' * 50)
```

```python
# Relational operator
print('pune' == 'Mumbai')
print('pune' == 'pune')
print('pune' != 'Pune')
print('pune' > 'Pune')
```

```python
# Logical operator

'hello' and 'world'
'hello' or 'world'
```

```python
# Loop
for i in 'hello':
  print(i)
```

```python
for i in 'hello':
  print('Pune')
```

```python
'P' in 'Pune'
```

```python
'P' not in 'Pune'
```

```python
# String Functions
# len
print(len('hello world'))
# min
print(min('hello world'))
# max
print(max('hello world'))
# sorted
print(sorted('hello world')) # sorted function retuen should be in list
print(sorted('hello world',reverse = True))
```

```python
# Capitalize/Title/Upper/Lower

s = 'Hello World'
print(s.capitalize())
print(s.title())
print(s.upper())
print(s.lower())
print(s.swapcase())
```

```python
# count/find

s = 'Hello world'
print(s.count('h'))

print(s.find('o'))

print(s.find('k'))

print(s.find('or'))
# Index() is similar as find() except if given character is not present in string it will throw error
print(s.index('d'))

#print(s.index('k'))

s.isalnum()

```
