```python
# How lists are store in memory

a = 2

print(a)
print(id(2))
print(id(a))

L = [1,2,3]
print(id(L))
print(id(L[0]))
print(id(L[1]))
print(id(L[2]))
print(id(1))
print(id(2))
print(id(3))
```

```python
# Charctestics
  # Orderd
  # Changable/Mutable
  # Hetrogenous
  # Can Be duplicate
  # dynamic
  # Can be nested
  # Can contain any kind of object
```

```python
# Orderd
L = [1,2,3]
L1 = [3,2,1]
print(L==L1)
```

```python
 # Hetrogenous
L = [1,5.6,10,[10,8,7],True]
print(L)
```

```python
# Creating List
# empty
L = []
print(L)

# 1D
L = [1,2,5,4,6,8,9]
print(L)
# 2D
L = [1,2,3,[4,5,6]]
print(L)

L = [1,True,53.6,5+7j,'Hello']
print(L)

# Type conversion
L = list('Hello')
print(L)
```

```python
# Acessing items from lis
# Indexing
L = [1,2,3,4,5]
print(L[0])
print(L[3])
#print(L[30])

# negative Indexing
print(L[-1])
print(L[-4])

L = [1,2,3,[4,5]]
print(L)
print(L[3])
print(L[3][0])
print(L[-1])
print(L[-1][-2])
```

```python
# Slicing

L = [1,2,3,4,5,6]
print(L[0:3])
print(L[-3:])
print(L[0::2])
print(L[-5:-2:2])
print(L[::-1])
```

```python
# Adding element in to the list
L = [1,2,3,4,5]
print(L)
print(id(L))
L.append(10)
print(L)
print(id(L))
L.append(True)
print(L)
print(id(L))
# extend
L.extend([6,7,8])
print(L)
print(id(L))
L.extend('Pune')
print(L)
print(id(L))
L.append('Pune')
print(L)
print(id(L))

# Insert
L = [1,2,3,4,5]
print(L)
L.insert(1,100)
print(L)
```

```python
# Editing items in list

L = [1,2,3,4,5,6]
print(L)
L[-1]=600
print(L)
L[1:4]=[200,300,400]
print(L)
```

```python
# Deleting items from list

L = [1,2,3,4,5]
print(L)
#del L
print(L)
del L[-1]
print(L)
del L[1:3]
print(L)

# remove
L = [1,2,3,4,5]
print(L)
L.remove(3)
print(L)
```

```python
# clear
L = [1,2,3,4,5]
print(L)
L.clear()
print(L)
```

```python
# pop
L = [1,2,3,4,5]
print(L)
L.pop(0)
print(L)
L.pop(3)
print(L)
L.pop(1)
print(L)
print(type(L))

a = 'hello'
print(type(a))
```
