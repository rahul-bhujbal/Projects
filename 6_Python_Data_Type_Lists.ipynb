```python
# Creating List
# empty
L = []
print(L)

# 1D
L = [1,2,5,4,6,8,9]
print(L)
# 2D
L = [1,2,3,[4,5,6]]
print(L)

L = [1,True,53.6,5+7j,'Hello']
print(L)

# Type conversion
L = list('Hello')
print(L)
```

```python
# Acessing items from lis
# Indexing
L = [1,2,3,4,5]
print(L[0])
print(L[3])
#print(L[30])

# negative Indexing
print(L[-1])
print(L[-4])

L = [1,2,3,[4,5]]
print(L)
print(L[3])
print(L[3][0])
print(L[-1])
print(L[-1][-2])
```

```python
# Slicing

L = [1,2,3,4,5,6]
print(L[0:3])
print(L[-3:])
print(L[0::2])
print(L[-5:-2:2])
print(L[::-1])
```

```python
# Adding element in to the list
L = [1,2,3,4,5]
print(L)
print(id(L))
L.append(10)
print(L)
print(id(L))
L.append(True)
print(L)
print(id(L))
# extend
L.extend([6,7,8])
print(L)
print(id(L))
L.extend('Pune')
print(L)
print(id(L))
L.append('Pune')
print(L)
print(id(L))

# Insert
L = [1,2,3,4,5]
print(L)
L.insert(1,100)
print(L)
```

```python
# Editing items in list

L = [1,2,3,4,5,6]
print(L)
L[-1]=600
print(L)
L[1:4]=[200,300,400]
print(L)
```

```python
# Deleting items from list

L = [1,2,3,4,5]
print(L)
#del L
print(L)
del L[-1]
print(L)
del L[1:3]
print(L)

# remove
L = [1,2,3,4,5]
print(L)
L.remove(3)
print(L)
```

```python
# clear
L = [1,2,3,4,5]
print(L)
L.clear()
print(L)
```

```python
# pop
L = [1,2,3,4,5]
print(L)
L.pop(0)
print(L)
L.pop(3)
print(L)
L.pop(1)
print(L)
print(type(L))

a = 'hello'
print(type(a))
```

```python
# Operations on List

# Arithmatic operation(+,*)

L1 = [1,2,3,4,5]
L2 = [6,7,8,9]
print(L1+L2)# merge

print(L1*3)

# Membership
L1 = [1,2,3,4,5]
L2 = [6,7,8,9]

print(5 in L1)
print(6 in L1)
print([4,5] in L1)
```

```python
# Loop
L1 = [1,2,3,4]
L2 = [1,2,3,4,[5,6]]

for i in L1:
  print(i)

for i in L2:
  print(i)
```

```python
# count()
L = [1,2,3,1,2,4,1,5]
print(L.count(1))
print(L.count(2))
```

```python
# index()
L = [1,2,3,1,2,4,1,5]
print(L.index(5))
print(L.index(1))
print(L.index(1,5))
L.reverse() # permanently reverse the list
print(L)
print(L.index(5))
```

```python
# List Comprehension
# List Comprehension provide a easy way of creating list
# newlist = [expression for itme in itrable if condition == True]

# Add 1 to 10 numbers in list
L = []
for i in range(1,11):
  L.append(i)
print(L)

L = [ i for i in range(1,11)]
print(L)

v = [2,3,4]
s = -3
print(v*s)
# output [-6,-9,-12]

x = []
for i in v:
  x.append(i*s)
print(x)

x = [s*i for i in v]
print(x)

# Add square
L = [1,2,3,4,5]
x = []
for i in L:
  x.append(i*i)
print(x)

x = [i*i for i in L]
print(x)

# print all numbers divisible by 5 in the range of 1 to 100

L = []
for i in range(1,101):
  if i%5==0:
    L.append(i)
print(L)


L = [i for i in range(1,101) if i%5==0]
print(L)

# Cartesion product
L1 = [1,2,3,4]
L2 = [5,6,7,8]
c = []
for i in L1:
  for j in L2:
    c.append(i*j)
print(c)

c = [i*j for i in L1 for j in L2]
print(c)

# Advantage of List comperhension
  # More time efficient and space efficient than loop
  # require fewer lines of code
  # Tranform itrative statement in formula
```
