```python
# Accessing item

my_dict = {'name':'Rahul', 'age':36, 'gender':'Male'}
print(my_dict['name'])
print(my_dict['gender'])


s = {
    'name': 'Rahul',
    'College': 'Bit',
    'Sem' : 4,
    'Subjects':{
        'dsa': 40,
        'math': 60,
        'englis': 55
    }
}

print(s)
print(s['Subjects'])
print(s['Subjects']['math'])

# get
my_dict = {'name':'Rahul', 'age':36, 'gender':'Male'}
print(my_dict['name'])
print(my_dict['gender'])

print(my_dict.get('name'))
print(my_dict.get('age'))

```

```python
# Adding key value pair

info = {'name':'Rahul','age':36}
print(info)
info['gender']='Male'
print(info)


s = {
    'name': 'Rahul',
    'College': 'Bit',
    'Sem' : 4,
    'Subjects':{
        'dsa': 40,
        'math': 60,
        'englis': 55
    }
}

print(s)
s['Subjects']['Python'] = 76
print(s)
```

```python
# Remove key value pair
my_dict = {'name':'Rahul', 'age':36, 'gender':'Male'}
print(my_dict)
my_dict.pop('gender')
print(my_dict)

my_dict = {'name':'Rahul', 'age':36, 'gender':'Male','weight':62}
print(my_dict)
my_dict.popitem()
print(my_dict)

my_dict.clear()
print(my_dict)

```

```python
# Editing
s = {
    'name': 'Rahul',
    'College': 'Bit',
    'Sem' : 4,
    'Subjects':{
        'dsa': 40,
        'math': 60,
        'englis': 55
    }
}

print(s)
s['Sem'] = 5
print(s)
s['Subjects']['dsa'] = 80
print(s)
```
