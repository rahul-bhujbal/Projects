1. Python output

2. Data Type

```python
# Integer
print(8)
# 1*10^308
print(1e308)

# Decimal/Float
print(4.55)
print(1.4e308)

# Boolean
print(True)

# Text/String

print('Hello world')

# Complax
print(5+6j)

# List
print([1,2,3,4,5])

# Tuple
print((1,2,3,4,5))

# set
print({1,2,3,4})

# Dictionary
print({'name':'Rahul','gender':'Male','Weight':62})
```

3. Variables

```python
# Static vs Dynamic Typing

name = 'Rahul'
print(name)
print(type(name))

a = 5
b = 6
print(type(a))

# static vs Dynamic Binding

a = 5
print(a)
print(type(a))

a = 'Mukesh'
print(a)
print(type(a))

# Stylish declaration
a = 1
b = 2
c = 3
print(a,b,c)
a,b,c=1,2,3
print(a,b,c)

a=b=c=5
print(a,b,b)
```
