```python
# *args and **kwargs

# *args and **kwargs are special python keywords thar are used to pass the varaible length argument to a function

```

```python
# *args
# allow us to pass a varaible number of non-keyword argument
```

```python
def multiplay(*args):
  product = 1
  for i in args:
    product = product * i
  return product
```

```python
multiplay(10,5,2,3)
```

```python
# How *args work, internaly it create a tuple
# print function also internally use *args

def multiplay(*args):
  product = 1
  for i in args:
    product = product * i
  print(args)
  return product
```

```python
multiplay(1,2,3,4,5,6,7,8,1,10,11)
```

```python
# **kwargs
# **kwargs allow us to pass any number of keyword argument
# Keyword argument means that they contain a key value pair, like a python dictionary


def show(**kwargs):
  for (key,value) in kwargs.items():
    print(key,' --> ',value)
```

```python
show(india='Delhi',Shrilanka = 'Colombo',Pak='Karachi')
```

```python
# Lambda Function
# A lambda function is a small anonymous function
# A lambda function can take any number of argument, but can only have one expression
```

```python
a = lambda x:x**2
print(a(2))

```

```python
print(a(3))
```

```python
add = lambda x,y: x+y
print(add(4,5))
```
