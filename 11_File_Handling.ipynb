```python
f = open('sample_4.txt','r')
while f.readline() != '':
  print(f.readline())
f.close()
```

```python
import csv
headr = ['Name','Age','Subject']
rows = [
    ['Rahul',35,'Python'],
    ['Ankita',30,'Snowflake'],
    ['Andy',37,'Java']
]

with open('students.csv','w') as files:
  writer = csv.writer(files)
  writer.writerow(headr)
  writer.writerows(rows)
```
