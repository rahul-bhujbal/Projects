Loops In Python


*   While loop
*   for loop


*   Nested for loop






```python
# While loop example- print the table

num = int(input('Enter the number'))
i = 1
while i < 11:
  print(num * i)
  i = i+1
```

```python
# while loop with else
num = int(input('Enter the number'))
i = 1
while i < 11:
  print(num * i)
  i = i+1
else:
  print('While condition become false, hence outside loop')
```

```python
# gussing game
# generate a random integer between 1 and 100

import random
jackpot = random.randint(1,100)
guess = int(input('Please enter jackpot number: '))
counter = 1
while guess != jackpot:
  if guess < jackpot:
    print('Wrong guess!! Please enter higher number')
  else:
    print('Wrong guess!! Please enter lower number')
  guess = int(input('Please enter jackpot number: '))
  counter = counter + 1
else:
  print('Congratulatio!! Your guess is correct at', counter,'round')
```

```python
# for loop
num = int(input('Enter number'))
for i in range(1,11):
  print(i*num)
```

```python
#
rows = int(input('Enter no of rows'))
for i in range(1,rows+1):
  for j in (range(1,i+1)):
    print('*',end='')
  print()
```

```python
rows = int(input('Enter number: '))
for i in range(1,rows+1):
  for j in range(1,i+1):
    print(j, end = '')
  print()
```

Loop Control Statement


*   Break
*   Countinue

*   pass






```python
for i in range(1,10):
  if i == 5:
    break
  print(i)
```

```python
for i in range(1,10):
  if i == 5:
    continue
  print(i)
```

```python
for i in range(1,10):
  pass
```
