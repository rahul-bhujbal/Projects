```python
# Decorators are flexible way to modify or extend behavior of function or method.
  # A decorator is a function that take another function as an argument and return a new function with ehanced funcationality
```

```python
def decorator(fun):
  def wrapper():
    print('Before calling')
    fun()
    print('After calling')
  return wrapper

@decorator
def xyz():
  print('Hello World')
xyz()
```

```python
# Decorator With Parameter
def decorator(fun):
  def wrapper(*args, **kwargs):
    print('Before calling')
    result = fun(*args, **kwargs)
    print('After calling')
    return result
  return wrapper

@decorator
def add(a,b):
  return a+b

print(add(2,3))
```

```python
# Higher-Order Function
  # Its functions that either take other function as input

def fun(f,x):
  return(f(x))

def square(x):
  return x*x

res = fun(square,3)
print(res)
```

```python
def sqaure(x):
  return x**2

def transform(f,L):
  out = []
  for i in L:
    out.append(f(i))
  print(out)

L = [1,2,3,4,5]
print(transform(sqaure,L))
```

```python
# Map
L = [1,2,3,4,5]
list(map(lambda x: x**2,L))
```

```python
L = [1,2,3,4,5]
list(map(lambda x:'even' if x%2==0 else 'odd',L))
```

```python
# filter
L = [1,2,3,4,5]
list(filter(lambda x: x%2==1,L))
```

```python
# Built in Decorators

class xyz:
  @staticmethod
  def add(x,y):
    return x+y

result = xyz.add(8,9)
print(result)
```

```python
# Generator

def simple_generator():
    for i in range(1, 6):
        yield i
# Create a generator object
gen = simple_generator()
# Iterate through the generator
for value in gen:
    print(value)
```

```python
def fun():
  yield 1
  yield 2
  yield 3
  yield 4


for i in fun():
  print(i)

```
