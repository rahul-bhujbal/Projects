4. Operators


*   Arthmetic

*   Relational


*   Logical
*   Bitwise

*   Assignment

```python
# Arthmetic
print(5+6)
print(6-5)
print(5*6)
print(5/2)
print(5//2)
print(5%2)
print(5**2)
```

```python
# Relational operator
print(5>6)
print(5<6)
print(5>=6)
print(5<=6)
print(5==6)
print(5!=6)
```

```python
# Logical
print (1 and 0)
print (1 or 0)
print(not 0)
```

```python
# Bitwise
print (bin(2  &  3))
print (2  &  3)
print (bin(2  |  3))
print(2  |  3)

print(2^3)
print(~3)
print(4>>2)
print(4<<2)
```

```python
# Assignment
a = 2
print(a)

a += 2

print(a)
a = a+2
print(a)

++a and a++
```
