```python
# Catching the specific error
try:
  with open('exception.txt','r') as f:
    print(f.read())
    #print(m)
    print(5/2)
    L = [1,2,3,4]
    L[1]
except FileNotFoundError as e:
  print(e.args)
except NameError as e:
  print(e.args)
except ZeroDivisionError as e:
  print(e.args)
except Exception as e:
  print(e.args)
```

```python
# Finally

try:
  with open('exception.txt','r') as f:
    print(f.read())
    #print(m)
    print(5/2)
    L = [1,2,3,4]
    L[1]
except NameError as e:
  print(e.args)
except FileNotFoundError as e:
  print(e.args)
except ZeroDivisionError as e:
  print(e.args)
except Exception as e:
  print(e.args)
finally:
  print('Always execute finally ')
```

```python
# Advantage of raise exception
class Bank:
  def __init__(self,balance):
    self.balance = balance
  def withdraw(self,amonut):
    if amonut < 0:
      raise Exception('Amount cannot be -ve')
    if self.balance < amonut:
      raise Exception('Insufficient Balance')
    self.balance = self.balance - amonut
    return self.balance

obj = Bank(50000)
try:
  obj.withdraw(-20000)
except Exception as e:
  print(e.args)
finally:
  print(obj.balance)
```
