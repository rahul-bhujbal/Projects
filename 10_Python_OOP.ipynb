```python
# We can implement method overloading using this way

class shape:
  def area(self,a,b=0):
    if b == 0:
      return 3.14*a*a
    else:
      return a*b
```

```python
s = shape()
print("Area of circle ",s.area(4))
print("Area of rectangle ",s.area(4,5))
```
