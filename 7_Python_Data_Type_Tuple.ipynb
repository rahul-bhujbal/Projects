```python
# Tuple
  # A Tuple is similar to a list. The diff is that we can not change the elements of a tuple once it has assigned
    # where we can change the element of list
  # In short, a tuple is an immutable list. A tuple can not be changed in any way once it ic created

# characterstics
  # orded
  # Unchangable
  # allow duplicate
# Plan of attack
  # Creating Tuple
  # Accessing Item
  # Editing Item
  # Adding Item
  # Deleting Item
  # Operations on tuple
  # Functions of Tuple
```

```python
# Creating Tuple
  # empty
t1 = ()
print(t1)

# Creating tuple with single item
t2 = (2)
print(t2)
print(type(t2))

t3 = ('hello')
print(t3)
print(type(t3))

t4 = (1,2,3,4)
print(t4)
print(type(t4))

t5 = (2,)
print(t5)
print(type(t5))

t6 = ('hello',)
print(t6)
print(type(t6))

t7 = (1,2,3,(6,8))
print(t7)

# type conversion
t8 = tuple(('hello'))
print(t8)
```

```python
# Indexing
t1 = (1,2,3)
print(t1)
print(t1[0])
print(t1[-1])
#print(t1[-6])
```

```python
# Slicing

t = (1,2,3,4)
print(t[0:4])
print(t[0:4:2])
print(t[-3:-1])
print(t[::-1])


t5 = (1,2,3,(4,5))
print(t5)
print(t5[-1][0])
```

```python
# Editing the items, Not able to edit it becouse its a immutable data type
t = (1,2,3,4)
print(t)
t[0]=100
print(t)
```

```python
# Adding items
t3 = (1,2,3,4)
print(t3)
# t3.append(5)
# Not possible becouse its immutable
```

```python
# Deleting items
t9 = (1,2,3)
print(t9)
del t9
# print(t9)

t10 = (1,2,3,4,5)
# del t10[-1]
```

```python
# Operation on tuple

# + and *

t1 = (1,2,3,4)
t2 = (4,5,6,7)
print(t1+t2)
print(t1*3)

# Membership
print(1 in t1)
print(6 in t1)
print(6 not in t2)

for i in t1:
  print(i)
```

```python
# Tuple functions
# len/min/max/sorted/sum
t = (8,7,23,6,89,45)
print(len(t))
print(min(t))
print(max(t))
print(sum(t))
print(sorted(t))

# count

print(t.count(23))

# Index
print(t.index(7))
#print(t.index(36))
```

```python
# Difference between List and Tuple
# Syntax
# Mutablity
# speed
# Memory
# Build in funcationality
# Error prone
# Usability

# speed

import time
L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*6
print('List Time',time.time()-start)

start = time.time()
for i in T:
  i*6
print('Tuple Time',time.time()-start)
```

```python
# Memory

import sys
L = list(range(1000))
T = tuple(range(1000))
print('List Size',sys.getsizeof(L))
print('Tuple Size',sys.getsizeof(T))

```

```python
# Error Prone
a = [1,2,3]
b = a
print(a)
print(b)
a.append(10)
print(a)
print(b)

a = (1,2,3)
b = a
a = a+(4,)
print(a)
print(b)
```

```python
#Tuple Unpacking
a,b,c = (1,2,3)
print(a,b,c)

a = 1
b = 2
a,b = b,a
print(a,b)


t = (1,2,3,list[4,5])
print(t)
print(type(t))
print(t)

a,b,*others=(1,2,3,4)
print(a,b)
print(*others)
```
