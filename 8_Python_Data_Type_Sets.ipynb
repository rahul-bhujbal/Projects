```python
# Sets
  # A sets is an unorded collection of items. Every set element is unique(no duplicate) and must be mutable
  # A set can also used to perform mathmatical operations like union, intersection etc

  # Charaterstics
    # Unorded
    # No duplicate
    # Cant contain mutable data type
```

```python
# Creating sets

# empty set

s = {}
print(s)
print(type(s))

s1 = {1,2,3}
print(s1)
print(type(s1))

#s2 = {1,2,3,{4,5}}
#print(s2)
#s3 = {1,2,3,[4,5]}
#print(s3)

s4 = {1,2,3,(4,5)}
print(s4)

s5 = {1,2,3,2,1,5}
print(s5)

# herto

s6 = {1,'hello',4.5,True}
print(s6)

# Type conversion

s7 = set([1,2,3])
print(s7)

# Duplicate not allowed
s8 = {1,1,2,2,3,3}
print(s8)

# Unorderd
S1 = {1,2,3}
s2 = {3,2,1}
print(s1==s2)
```

```python
# Accessing items does not work becouse its unorderd

s1 = {1,2,3,4,5}
print(s1)
print(s1[0])
```

```python
# Editing
s3 = {1,2,3}
s3[0] = 100
print(s3)
```

```python
# Adding items
s = {1,2,3,4}
print(s)
s.add(50)
print(s)

s1 = {1,2,3,4}
print(s1)
s1.update([9,8,7])
print(s1)
```

```python
# Deleting

s = {1,2,3,4,5}
print(s)
#del s
#print(s)

#del s[0]
# discard
s.discard(5)
print(s)

s.discard(60) # not throw error
print(s)

# remove
s1 = {1,2,3,4,5}
print(s1)
s1.remove(5)
print(s1)

# s1.remove(60)# it will throw error

# pop
s3 = {1,2,3,4,5}
print(s3)
s3.pop()
print(s3)

s4 = {1,2,3,4,5}
print(s4)
s4.pop()
print(s4)
```

```python
# clear
s4 = {1,2,3,4,5}
print(s4)
s4.clear()
print(s4)
```

```python
# set operation

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# union (|)
print(s1|s2)

# Intersection
print(s1&s2)
# Difference
print(s1-s2)
print(s2-s1)

# Membership

print(1 in s1)
print(1 not in s1)

# Itration

for i in s1:
  print(i)
```

```python
# Set functions
# len/sum/min/max/sorted
s = {3,1,4,5,2,7,8}
print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s))
```

```python
# issubset/issuperset/isdisjoint
# isdisjoint
s1 = {1,2,3,4}
s2 = {3,4,5,5}
print(s1.isdisjoint(s2)) # if all items are different in set then it will return True otherwise it will return false
s3 = {1,2,3,4}
s4 = {7,8,9,10}
print(s3.isdisjoint(s4))

# issubset
s5 = {1,2,3,4,5,6}
s6 = {3,4,5}

print(s5.issubset(s6))
print(s6.issubset(s5))

# issuperset
s7 = {1,2,3,4,5,6}
s8 = {3,4,5}
print(s7.issuperset(s8))
print(s8.issuperset(s7))


```

```python
# Frozenset

#s = {1,2,3,[4,5]}
#print(s)

fs = frozenset([1,2,3])
print(fs)
fs1 = frozenset((1,2,3))
print(fs1)
fs2 = frozenset({1,2,3})
print(fs2)
```
